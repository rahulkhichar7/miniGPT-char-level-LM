```python
import re
import requests
from pathlib import Path


OUTPUT_PATH = Path("data/input.txt")


def download_book(book_id):
    url = (
        f"https://www.gutenberg.org/cache/epub/"
        f"{book_id}/pg{book_id}.txt"
    )

    print(f"Downloading: {BOOKS[book_id]}")

    response = requests.get(url, timeout=60)
    response.raise_for_status()

    return response.content.decode("utf-8")


def clean_book(text):
    # Normalize line endings
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Remove Gutenberg header
    start = re.search(
        r"\*\*\* START OF (?:THE|THIS) PROJECT GUTENBERG EBOOK.*?\*\*\*",
        text,
        flags=re.IGNORECASE | re.DOTALL,
    )

    if start:
        text = text[start.end():]

    # Remove Gutenberg footer
    end = re.search(
        r"\*\*\* END OF (?:THE|THIS) PROJECT GUTENBERG EBOOK.*?\*\*\*",
        text,
        flags=re.IGNORECASE | re.DOTALL,
    )

    if end:
        text = text[:end.start()]

    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r" *\n *", "\n", text)
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


def main():

    OUTPUT_PATH.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    cleaned_books = []
    count = 0
    length = []

    for book_id in BOOKS:
        print(f"Count: {count}")
        count += 1

        try:
            raw_text = download_book(book_id)
            clean_text = clean_book(raw_text)

            cleaned_books.append(clean_text)
            length.append(len(clean_text))

            print(
                f"  → {len(clean_text):,} characters"
            )

        except Exception as e:
            print(
                f"  ERROR: {BOOKS[book_id]}: {e}"
            )

    # Separate books
    final_text = "\n\n\n".join(cleaned_books)

    OUTPUT_PATH.write_text(
        final_text,
        encoding="utf-8"
    )

    size_mb = (
        OUTPUT_PATH.stat().st_size / 1024 / 1024
    )

    print("\n" + "=" * 60)
    print("Dataset created")
    print("=" * 60)

    print(f"Books      : {len(cleaned_books)}")
    print(f"Characters : {len(final_text):,}")
    print(f"File size  : {size_mb:.2f} MB")
    print(f"Output     : {OUTPUT_PATH}")
    print(f"Max char in a book : {max(length)}")


if __name__ == "__main__":
    main()
```
